```python
import pandas as pd

from sklearn.model_selection import  train_test_split
```

```python
df = pd.read_excel('QSPR_data_thia_michael.xlsx')
df
```

```python
train_idx, test_idx = train_test_split(df, test_size=0.15, random_state=15)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Ustawienie stylu
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))

# Rysowanie rozkładu dla zbioru treningowego
sns.kdeplot(train_idx['ee'], label='Train (ee)', fill=True, color='blue', alpha=0.3)

# Rysowanie rozkładu dla zbioru testowego
sns.kdeplot(test_idx['ee'], label='Test (ee)', fill=True, color='orange', alpha=0.3)

# Dodanie detali
plt.title('Porównanie reprezentatywności cechy "ee"', fontsize=15)
plt.xlabel('Wartość cechy ee')
plt.ylabel('Gęstość (Density)')
plt.legend()

plt.show()
```

```python
import pandas as pd

# Przygotowanie danych do boxplota
df_plot = pd.concat([
    pd.DataFrame({'value': train_idx['ee'], 'set': 'Train'}),
    pd.DataFrame({'value': test_idx['ee'], 'set': 'Test'})
])

plt.figure(figsize=(8, 5))
sns.boxplot(x='set', y='value', data=df_plot, palette='Set2')
plt.title('Porównanie statystyk cechy "ee"')
plt.show()
```

```python
num_bins = 5
df['ee_bin'] = pd.cut(df['ee'], bins=num_bins, labels=False)
train_idx, test_idx = train_test_split(df, test_size=0.15, stratify=df['ee_bin'], random_state=42)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Ustawienie stylu
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))

# Rysowanie rozkładu dla zbioru treningowego
sns.kdeplot(train_idx['ee'], label='Train (ee)', fill=True, color='blue', alpha=0.3)

# Rysowanie rozkładu dla zbioru testowego
sns.kdeplot(test_idx['ee'], label='Test (ee)', fill=True, color='orange', alpha=0.3)

# Dodanie detali
plt.title('Porównanie reprezentatywności cechy "ee"', fontsize=15)
plt.xlabel('Wartość cechy ee')
plt.ylabel('Gęstość (Density)')
plt.legend()

plt.show()
```
