```python
import pandas as pd

from sklearn.model_selection import  train_test_split
```

```python
df = pd.read_excel('QSPR_epoxidation_ee.xlsx')
df
```

```python
df['epox_cla'].hist(bins=5)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))

# Get counts, bin edges and bar patches
counts, bins, patches = plt.hist(
    df['epox_cla'],
    bins=[-0.5, 0.5, 1.5],
    color='tab:blue',
    edgecolor='black',
    alpha=0.8
)

plt.xticks([0, 1], ['0', '1'])
plt.xlabel('epox_cla', fontsize=30)
#plt.ylabel('Count')
#plt.title('Histogram of epox_cla (0/1)')
plt.xticks(fontsize=26)
plt.yticks([0, 20, 40], ['0', '20', '40'], fontsize=26)
plt.tight_layout()

# Add count labels on top of each bar
for c, p in zip(counts, patches):
    x = p.get_x() + p.get_width() / 2
    y = p.get_height() - 10
    plt.text(x, y, str(int(c)), ha='center', va='bottom', fontsize=26)

plt.savefig("epox_cla_hist_larger.svg", format="svg")
plt.show()
```

```python
train_idx, test_idx = train_test_split(df, test_size=0.15, random_state=15, stratify=df['epox_cla'])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Global style for publication-quality plots
sns.set_theme(style="whitegrid", font_scale=1.4)  # larger fonts for print/PDF[web:35][web:41]

plt.figure(figsize=(8, 5))

# KDE for training set
sns.kdeplot(
    data=train_idx['epox_cla'],
    label='Train (epox_cla)',
    fill=True,
    color='tab:blue',
    alpha=0.3,
    linewidth=2
)

# KDE for test set
sns.kdeplot(
    data=test_idx['epox_cla'],
    label='Test (epox_cla)',
    fill=True,
    color='tab:orange',
    alpha=0.3,
    linewidth=2
)

# Titles and labels in English
plt.title('Comparison of the Representativeness of the "epox_cla" Feature', fontsize=16)
plt.xlabel('epox_cla value', fontsize=14)
plt.ylabel('Density', fontsize=14)

# Legend and layout tweaks
plt.legend(frameon=False, fontsize=12)
plt.tight_layout()

# Optional: save as vector graphic for publication
plt.savefig("epox_cla_kde_train_test.svg", format="svg", dpi=300)  # high‑resolution, vector output[web:13][web:18]

plt.show()
```
