```python
import pandas as pd

from sklearn.model_selection import  train_test_split
```

```python
df = pd.read_excel('new_data_epoxidation.xlsx')
df
```

```python
df['ee'].hist(bins=5)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))

# Histogram for continuous ee values
counts, bins, patches = plt.hist(
    df['ee'],
    bins=5,                 # adjust number of bins as needed
    color='tab:blue',
    edgecolor='black',
    alpha=0.8
)

plt.xlabel('ee value', fontsize=40)
#plt.ylabel('Count')
#plt.title('Histogram of ee')
plt.xticks(fontsize=36)
plt.yticks([0, 15, 30], ['0', '15', '30'], fontsize=36)
plt.tight_layout()

# Add count labels on top of each bar
for c, p in zip(counts, patches):
    if c == 0:
        continue  # skip empty bins to avoid clutter
    x = p.get_x() + p.get_width() / 2
    y = p.get_height() -6
    plt.text(x, y, f'{int(c)}', ha='center', va='bottom', fontsize=36)

# Optional: save as SVG for publication
plt.savefig("ee_histogram.svg", format="svg", dpi=300)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Global style for publication-quality plots
sns.set_theme(style="whitegrid", font_scale=1.4)  # larger fonts for print/PDF[web:35][web:41]

plt.figure(figsize=(8, 5))

# KDE for training set
sns.kdeplot(
    data=train_idx['ee'],
    label='Train (ee)',
    fill=True,
    color='tab:blue',
    alpha=0.3,
    linewidth=2
)

# KDE for test set
sns.kdeplot(
    data=test_idx['ee'],
    label='Test (ee)',
    fill=True,
    color='tab:orange',
    alpha=0.3,
    linewidth=2
)

# Titles and labels in English
plt.title('Comparison of the Representativeness of the "ee" Feature', fontsize=16)
plt.xlabel('ee value', fontsize=14)
plt.ylabel('Density', fontsize=14)

# Legend and layout tweaks
plt.legend(frameon=False, fontsize=12)
plt.tight_layout()

# Optional: save as vector graphic for publication
#plt.savefig("ee_kde_train_test.svg", format="svg", dpi=300)  # high‑resolution, vector output[web:13][web:18]

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Prepare data for boxplot (long format)
df_plot = pd.concat([
    pd.DataFrame({'value': train_idx['ee'], 'set': 'Train'}),
    pd.DataFrame({'value': test_idx['ee'],  'set': 'Test'})
], ignore_index=True)

# Publication-style aesthetics
sns.set_theme(style="whitegrid", font_scale=1.4)  # good defaults for papers[web:51][web:55]

plt.figure(figsize=(6, 4))

ax = sns.boxplot(
    x='set',
    y='value',
    data=df_plot,
    palette='Set2',
    width=0.5,
    fliersize=3,          # smaller outlier markers
    linewidth=1.5
)

ax.set_title('Comparison of "ee" Feature Statistics', fontsize=16)
ax.set_xlabel('Data set', fontsize=14)
ax.set_ylabel('ee value', fontsize=14)

sns.despine(offset=5, trim=True)  # cleaner axes[web:51]
plt.tight_layout()

# Optional: save as high-quality SVG for publication
#plt.savefig("ee_boxplot_train_test.svg", format="svg", dpi=300)

plt.show()
```
